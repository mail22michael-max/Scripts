## 3. List Mutability (Changing Lists)

Unlike strings, **lists are mutable** - you can change elements after creation!

### Changing Individual Elements:

````python
# Create a list
my_list = ['a', 'b', 'c', 'd', 'e']
print("Original:", my_list)
````

````python
# Change the first element
my_list[0] = 'A'
print("After change:", my_list)
````

````python
# Change multiple elements at once
my_list[1:3] = ['B', 'C']
print("After multiple changes:", my_list)
````

### Comparison with Strings (Immutable):

````python
# ✅ Lists are mutable
my_list = [1, 2, 3]
my_list[0] = 999
print("List modified:", my_list)
````

````python
# ❌ Strings are immutable - this would cause an error!
# my_string = "hello"
# my_string[0] = 'H'  # TypeError!
````

## 4. List Methods

Lists have many built-in methods for adding, removing, and modifying elements.

### append() - Add to End

Adds an element to the **end** of the list.

````python
# Start with a list
fruits = ['apple', 'banana']
print("Original:", fruits)
````

````python
# Append an item
fruits.append('orange')
print("After append:", fruits)
````

````python
# Append another item
fruits.append('grape')
print("After another append:", fruits)
````

### insert() - Add at Specific Position

Inserts an element at a **specific index**.

````python
# Insert at index 1
fruits.insert(1, 'mango')
print("After insert:", fruits)
````

````python
# Insert at the beginning (index 0)
fruits.insert(0, 'strawberry')
print("After insert at beginning:", fruits)
````

### extend() - Add Multiple Items

Adds all items from another list.

````python
# Extend with another list
more_fruits = ['kiwi', 'melon']
fruits.extend(more_fruits)
print("After extend:", fruits)
````

### remove() - Remove by Value

Removes the **first occurrence** of a value.

````python
# Remove 'banana'
fruits.remove('banana')
print("After remove:", fruits)
````

### pop() - Remove by Index

Removes and **returns** an element at a specific index (default is last item).

````python
# Start fresh
numbers = [1, 2, 3, 4, 5]
print("Original:", numbers)
````

````python
# Pop the last item
popped = numbers.pop()
print("Popped item:", popped)
print("List after pop:", numbers)
````

````python
# Pop item at index 1
popped = numbers.pop(1)
print("Popped item:", popped)
print("List after pop:", numbers)
````

### clear() - Remove All Items

````python
# Create a list
temp_list = [1, 2, 3, 4, 5]
print("Before clear:", temp_list)
````

````python
# Clear all items
temp_list.clear()
print("After clear:", temp_list)
````

### index() - Find Position of Value

````python
# Find the index of 'orange'
fruits = ['apple', 'banana', 'orange', 'grape']
position = fruits.index('orange')
print("Index of 'orange':", position)
````

### count() - Count Occurrences

````python
# Count how many times 'apple' appears
fruits = ['apple', 'banana', 'apple', 'orange', 'apple']
count = fruits.count('apple')
print("'apple' appears", count, "times")
````

## 5. List Operations

### Concatenation (+)

Combine two lists into a new list.

````python
# Concatenate lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]
combined = list1 + list2
print("Combined:", combined)
````

````python
# Original lists are unchanged
print("list1:", list1)
print("list2:", list2)
````

### Repetition (*)

Repeat a list multiple times.

````python
# Repeat a list
my_list = [1, 2, 3]
repeated = my_list * 3
print("Repeated:", repeated)
````

````python
# Create a list of zeros
zeros = [0] * 5
print("Zeros:", zeros)
````

### Membership Testing (in)

Check if an item is in a list.

````python
# Check if item exists
fruits = ['apple', 'banana', 'orange']
print('apple' in fruits)
print('grape' in fruits)
````

````python
# Check if item does NOT exist
print('grape' not in fruits)
````

## 6. Sorting and Reversing Lists

### sort() - Sort in Place

Sorts the list **permanently** (modifies the original list).

````python
# Sort numbers
numbers = [5, 2, 8, 1, 9, 3]
print("Before sort:", numbers)
numbers.sort()
print("After sort:", numbers)
````

````python
# Sort strings (alphabetically)
names = ['Charlie', 'Alice', 'Bob']
print("Before sort:", names)
names.sort()
print("After sort:", names)
````

````python
# Sort in descending order
numbers = [5, 2, 8, 1, 9, 3]
numbers.sort(reverse=True)
print("Sorted descending:", numbers)
````

### sorted() - Return Sorted Copy

Returns a **new sorted list** (doesn't modify the original).

````python
# Create sorted copy
numbers = [5, 2, 8, 1, 9, 3]
sorted_numbers = sorted(numbers)
print("Original:", numbers)
print("Sorted copy:", sorted_numbers)
````

### reverse() - Reverse in Place

Reverses the list **permanently**.

````python
# Reverse a list
my_list = [1, 2, 3, 4, 5]
print("Before reverse:", my_list)
my_list.reverse()
print("After reverse:", my_list)
````

### Slicing to Reverse (Temporary)

Creates a **reversed copy** without modifying the original.

````python
# Reverse using slicing
my_list = [1, 2, 3, 4, 5]
reversed_copy = my_list[::-1]
print("Original:", my_list)
print("Reversed copy:", reversed_copy)
````

## 7. Nested Lists (Lists within Lists)

Lists can contain other lists! This is useful for creating matrices or tables.

### Creating Nested Lists:

````python
# Create a matrix (2D list)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(matrix)
````

### Accessing Nested Elements:

Use **double indexing** `[row][column]`

````python
# Get first row
print("First row:", matrix[0])
````

````python
# Get first element of first row
print("Element [0][0]:", matrix[0][0])
````

````python
# Get element from row 1, column 2 (middle of second row)
print("Element [1][1]:", matrix[1][1])
````

````python
# Get last element of last row
print("Element [2][2]:", matrix[2][2])
````

### Modifying Nested Lists:

````python
# Change a value in the matrix
matrix[0][0] = 999
print("Modified matrix:", matrix)
````

### Practical Example - Student Grades:

````python
# List of students with their grades
students = [
    ['Alice', 85, 90, 92],
    ['Bob', 78, 85, 88],
    ['Charlie', 92, 95, 90]
]

# Access Alice's first grade
print("Alice's first grade:", students[0][1])

# Access Bob's name
print("Second student's name:", students[1][0])
````

## 8. Built-in Functions for Lists

### len() - Length of List

````python
my_list = [1, 2, 3, 4, 5]
print("Length:", len(my_list))
````

### max() and min() - Maximum and Minimum

````python
numbers = [5, 2, 8, 1, 9, 3]
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
````

### sum() - Sum of All Elements

````python
numbers = [1, 2, 3, 4, 5]
total = sum(numbers)
print("Sum:", total)
````

````python
# Calculate average
average = sum(numbers) / len(numbers)
print("Average:", average)
````

## 🎓 Summary

### What we learned:

1. **Creating Lists:**
   - Use square brackets: `[item1, item2, item3]`
   - Can contain different types
   - Use `len()` to get length

2. **Indexing and Slicing:**
   - Same as strings: `list[0]`, `list[-1]`
   - Slicing: `list[start:stop:step]`

3. **Mutability:**
   - Lists can be changed: `list[0] = 'new value'`
   - Unlike strings (immutable)

4. **List Methods:**
   - Add: `.append()`, `.insert()`, `.extend()`
   - Remove: `.remove()`, `.pop()`, `.clear()`
   - Info: `.index()`, `.count()`

5. **Operations:**
   - Concatenate: `list1 + list2`
   - Repeat: `list * 3`
   - Membership: `item in list`

6. **Sorting and Reversing:**
   - `.sort()` - sort in place
   - `sorted()` - return sorted copy
   - `.reverse()` - reverse in place
   - `[::-1]` - reversed copy

7. **Nested Lists:**
   - Lists within lists
   - Access with double indexing: `matrix[row][col]`

8. **Built-in Functions:**
   - `len()`, `max()`, `min()`, `sum()`
