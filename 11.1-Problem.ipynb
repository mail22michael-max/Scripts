Let's use `while` loops to create a guessing game.

The Challenge:

Write a program that picks a random integer from 1 to 100, and has players guess the number. The rules are:

1. If a player's guess is less than 1 or greater than 100, say "OUT OF BOUNDS"
2. On a player's first turn, if their guess is
 * within 10 of the number, return "WARM!"
 * further than 10 away from the number, return "COLD!"
3. On all subsequent turns, if a guess is 
 * closer to the number than the previous guess return "WARMER!"
 * farther from the number than the previous guess, return "COLDER!"
4. When the player's guess equals the number, tell them they've guessed correctly *and* how many guesses it took!

You can try this from scratch, or follow the steps outlined below. A separate Solution notebook has been provided. Good luck!

```python
import random


def get_guess():
    """Prompt the player for a guess, re-prompting on non-integer input."""
    while True:
        raw = input("Guess a number between 1 and 100: ").strip()
        try:
            return int(raw)
        except ValueError:
            print("Please enter a whole number.")


def play():
    number = random.randint(1, 100)
    valid_guesses = 0          # counts only in-bounds guesses
    previous_distance = None   # distance from the number on the prior valid guess

    print("I'm thinking of a number between 1 and 100. Try to guess it!")

    while True:
        guess = get_guess()

        if guess < 1 or guess > 100:
            print("Out of bounds")
            continue

        valid_guesses += 1
        distance = abs(guess - number)

        if guess == number:
            print(f"Correct! The number was {number}. It took you {valid_guesses} guesses.")
            break

        if previous_distance is None:
            # First valid turn
            print("Warm!" if distance <= 10 else "Cold!")
        else:
            # Subsequent turns
            if distance < previous_distance:
                print("Warmer!")
            elif distance > previous_distance:
                print("Colder!")
            else:
                print("Same distance as before!")

        previous_distance = distance


if __name__ == "__main__":
    play()
```
